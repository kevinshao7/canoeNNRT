```python
import numpy as np
from netCDF4 import Dataset
import os

# -----------------------------
# User inputs
# -----------------------------
ncfile = "uranus2s.out2.00026.nc"       # input NetCDF file
outdir = "ic_bin"        # output directory
vars_to_write = ["rho", "press", "vel1", "vel2", "vel3"]

os.makedirs(outdir, exist_ok=True)

# -----------------------------
# Open NetCDF
# -----------------------------
ds = Dataset(ncfile, "r")

# Read dimensions (skip time)
rho = ds.variables["rho"]
_, nx, ny, nz = rho.shape

print(f"nx={nx}, ny={ny}, nz={nz}")

# -----------------------------
# Write header
# -----------------------------
with open(os.path.join(outdir, "header.txt"), "w") as f:
    f.write(f"{nx} {ny} {nz}\n")
    f.write("float32\n")
    for v in vars_to_write:
        f.write(f"{v}.bin\n")

# -----------------------------
# Write each variable
# -----------------------------
for vname in vars_to_write:
    var = ds.variables[vname]

    # Extract time=0 slice → shape (nx, ny, nz)
    data = var[0, :, :, :]

    # Ensure float32 & C-order
    data = np.asarray(data, dtype=np.float32, order="C")

    # Write raw binary
    outpath = os.path.join(outdir, f"{vname}.bin")
    data.tofile(outpath)

    print(f"Wrote {outpath}  shape={data.shape}")

ds.close()

print("Conversion complete.")
```
